### DATA CLEANING

```python
movie_credits = movie_credits[["movie_id","title_x","overview","genres","keywords","cast","crew"]]
movie_credits.head()
```

```python
movie_credits.dropna(inplace = True)
```

```python
movie_credits.shape #only 3 rows was deleted  
```

```python
values = json.loads(movies["genres"][0])
values
```

```python
values = json.loads(movies["genres"][0])  #to just fetch names
[value["name"] for value in values]
```

```python
def extract_values(str_lst):
    values = json.loads(str_lst)
    return [value["name"] for value in values]
```

```python
extract_values(movies["genres"][0])
```

```python
movie_credits["genres"] = movie_credits["genres"].apply(extract_values)
```

```python
movie_credits["keywords"] = movie_credits["keywords"].apply(extract_values)
```

```python
movie_credits
```

```python
movie_credits["cast"].apply(extract_values)
```

```python
movie_credits["cast"] = movie_credits["cast"].apply(extract_values).apply(lambda x:x[:3]) 
#lambda function is for temporary use
#(x:x[:3])->to fetch starting 3 values from every list
```

```python
movie_credits["cast"]
```

```python
json.loads(movie_credits["crew"][0])
```

```python
def fetch_director(str_lst):
    values = json.loads(str_lst)
    return [value["name"] for value in values if value["job"] == "Director"]
```

```python
movie_credits["crew"] = movie_credits["crew"].apply(fetch_director)
```

```python
movie_credits
```

```python
movie_credits["overview"] = movie_credits["overview"].str.split()
```

```python
movie_credits.head()
```

```python
def collapse(lst):
    final_lst = []
    for i in lst:
        final_lst.append(i.replace(" ",""))
    return final_lst
```

```python
movie_credits["cast"] = movie_credits["cast"].apply(collapse)
movie_credits["genres"] = movie_credits["genres"].apply(collapse)
movie_credits["crew"] = movie_credits["crew"].apply(collapse)
movie_credits["keywords"] = movie_credits["keywords"].apply(collapse)
```

```python
movie_credits
```

```python
movie_credits["tags"] = movie_credits["overview"] + movie_credits["genres"] + movie_credits["keywords"] + movie_credits["cast"] + movie_credits["crew"] 
```

```python
movie_credits.head()
```

```python
final_df = movie_credits[["movie_id", "title_x", "tags"]]
final_df
```

```python
final_df["tags"] = final_df["tags"].apply(lambda x: " ".join(x))
```

```python
final_df.columns
```

```python
final_df.rename({"title_x" : "title"}, axis = 1, inplace = True)
```

```python
final_df
```

```python
final_df["tags"][0]
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()

ps.stem("love")    #already defined by algorithm
```

```python
def stem(txt):
    lst = []
    for i in txt.split():
        lst.append(ps.stem(i))
    return " ".join(lst)
```

```python
final_df["tags"] = final_df["tags"].apply(lambda x: x.lower())
```

```python
final_df["tags"] = final_df["tags"].apply(stem)
```

```python
print(final_df["tags"].iloc[0][:100])
```

```python
final_df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=5000, stop_words="english")
```

```python
vector = cv.fit_transform(final_df["tags"])
```

```python
vector.toarray()
```

```python
list(cv.get_feature_names_out()) #these are 5000 unique words used in ovweview of every movie 
```

```python
final_df.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity.shape
```

```python
similarity
```

```python
movie_index = final_df[final_df["title"] == "Iron Man 2"].index[0]
movie_index
```

```python
distances = similarity[movie_index]  #cosine similarity of "Iron Man 2" movie with other movies
distances
```

```python
similar_movie = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])
similar_movie
#when a movie is compared to itself cosine similarity is 1 
```

```python
for index, score in similar_movie[:10]:
    print(final_df.iloc[index, 1])
```

```python
def recommender(movie_name):
    movie_index = final_df[final_df["title"] == movie_name].index[0]
    distances = similarity[movie_index]
    similar_movie = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])
    
    for index, score in similar_movie[:10]:
        print(final_df.iloc[index, 1])
```

```python
recommender("Avatar")
```
